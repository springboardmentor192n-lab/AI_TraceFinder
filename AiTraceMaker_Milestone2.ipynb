# Scanner Fingerprint Extraction

```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

flatfield = base + "/Flatfield"

# Select scanner
scanner = os.listdir(flatfield)[0]

scanner_path = flatfield + "/" + scanner

residuals = []

print("Scanner:",scanner)

for file in os.listdir(scanner_path):

    if file.endswith(".tif"):

        path = scanner_path+"/"+file

        img = cv2.imread(path,0)

        if img is not None:

            blur = cv2.GaussianBlur(img,(5,5),0)

            residual = img - blur

            residuals.append(residual)

# Average residuals
fingerprint = np.mean(residuals,axis=0)

plt.imshow(fingerprint,cmap='gray')

plt.title("Scanner Fingerprint")

plt.axis("off")
```

```python
import cv2
import os
import numpy as np

flatfield = base + "/Flatfield"

fingerprints_150 = {}
fingerprints_300 = {}

for scanner in os.listdir(flatfield):

    scanner_path = flatfield+"/"+scanner

    residuals150=[]
    residuals300=[]

    for file in os.listdir(scanner_path):

        if file.endswith(".tif"):

            img=cv2.imread(scanner_path+"/"+file,0)

            if img is not None:

                blur=cv2.GaussianBlur(img,(5,5),0)

                residual=img-blur

                if "150" in file:
                    residuals150.append(residual)

                if "300" in file:
                    residuals300.append(residual)

    # Fingerprint 150dpi
    if len(residuals150)>0:

        fingerprint150=np.mean(residuals150,axis=0)

        fingerprints_150[scanner]=fingerprint150

        print("150dpi fingerprint:",scanner)


    # Fingerprint 300dpi
    if len(residuals300)>0:

        fingerprint300=np.mean(residuals300,axis=0)

        fingerprints_300[scanner]=fingerprint300

        print("300dpi fingerprint:",scanner)
```

```python
import cv2
import os

official = base + "/Official"

test_path = None

# Find first real image automatically
for root, dirs, files in os.walk(official):

    for file in files:

        if file.endswith(".tif") or file.endswith(".png") or file.endswith(".jpg"):

            test_path = os.path.join(root,file)

            break

    if test_path is not None:
        break

print("Test Image:",test_path)
```

```python
img = cv2.imread(test_path,0)

if img is None:
    print("Image not loaded")
else:

    blur = cv2.GaussianBlur(img,(5,5),0)

    test_residual = img - blur

    print("Residual created successfully")
```

```python
scores = {}

fingerprints = fingerprints_300   # or fingerprints_150

for scanner in fingerprints:

    fingerprint = fingerprints[scanner]

    resized = cv2.resize(test_residual,
                         (fingerprint.shape[1],
                          fingerprint.shape[0]))

    score = np.corrcoef(resized.flatten(),
                        fingerprint.flatten())[0,1]

    scores[scanner] = score


for s in scores:

    print(s,"→",scores[s])
```

```python
predicted = max(scores,key=scores.get)

print("\nPredicted Scanner:",predicted)
```

```python
import cv2
import os
import numpy as np

# Dataset path
base = "/content/drive/MyDrive/ScannerProjectRaw/Dataset/Flatfield"

X = []
y = []

print("Extracting Features...")

for scanner in os.listdir(base):

    scanner_path = base + "/" + scanner

    for file in os.listdir(scanner_path):

        if file.endswith(".tif"):

            path = scanner_path + "/" + file

            img = cv2.imread(path,0)

            if img is None:
                continue

            # Residual extraction
            blur = cv2.GaussianBlur(img,(5,5),0)

            residual = img - blur

            # Resize
            residual = cv2.resize(residual,(128,128))

            # Flatten
            feature = residual.flatten()

            X.append(feature)

            y.append(scanner)

X = np.array(X)

y = np.array(y)

print("Features Shape:",X.shape)

print("Labels Shape:",y.shape)
```

```python
import cv2
import os
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Dataset path
base = "/content/drive/MyDrive/ScannerProjectRaw/Dataset/Flatfield"

X = []
y = []

print("Extracting Features...")

for scanner in os.listdir(base):

    scanner_path = base + "/" + scanner

    for file in os.listdir(scanner_path):

        if file.endswith(".tif"):

            path = scanner_path + "/" + file

            img = cv2.imread(path,0)

            if img is None:
                continue

            # Residual extraction
            blur = cv2.GaussianBlur(img,(5,5),0)

            residual = img - blur

            # Resize
            residual = cv2.resize(residual,(128,128))

            # Flatten
            feature = residual.flatten()

            X.append(feature)

            y.append(scanner)


X = np.array(X)
y = np.array(y)

print("Features Shape:",X.shape)


print("Training Model...")

model = SVC()

model.fit(X,y)

pred = model.predict(X)

accuracy = accuracy_score(y,pred)

print("Accuracy:",accuracy)
```

```python
import cv2
import os
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Dataset path
base = "/content/drive/MyDrive/ScannerProjectRaw/Dataset/Flatfield"

X = []
y = []

print("Extracting Features...")

for scanner in os.listdir(base):

    scanner_path = base + "/" + scanner

    for file in os.listdir(scanner_path):

        if file.endswith(".tif"):

            path = scanner_path + "/" + file

            img = cv2.imread(path,0)

            if img is None:
                continue

            # Residual extraction
            blur=cv2.GaussianBlur(img,(5,5),0)

            residual=img-blur

            residual=cv2.resize(residual,(128,128))

            feature=residual.flatten()

            X.append(feature)

            y.append(scanner)


X=np.array(X)
y=np.array(y)

print("Features Shape:",X.shape)


print("\nTraining Model...")

model=SVC()

model.fit(X,y)

pred=model.predict(X)


# Accuracy
accuracy=accuracy_score(y,pred)

print("\nAccuracy:",accuracy)


# Precision Recall F1
print("\nClassification Report:\n")

print(classification_report(y,pred))


# Confusion Matrix
print("\nConfusion Matrix:\n")

print(confusion_matrix(y,pred))
```
